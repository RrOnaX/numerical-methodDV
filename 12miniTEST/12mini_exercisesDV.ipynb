```python
import numpy as np

M = np.arange(24).reshape(6, 4)

print("shape:", M.shape)
print("ndim:", M.ndim)
print("size:", M.size)
print("dtype:", M.dtype)
```

```python
b = M.reshape(4, 6)
c = b.reshape(2, 3, 4)
flat = c.reshape(24,)

assert np.array_equal(flat, np.arange(24))
print("значения не изменились:", np.array_equal(flat, np.arange(24)))
```

```python
col = M[:, 2]
col[0] = 999

print("срез после изменения:", col)
print("оригинал M:\n", M)

M[0, 2] = 2
```

```python
mask = M  > 10

print("элементы > 10:", M[mask])
print("среднее:", M[mask].mean())
```

```python
row_means = M.mean(axis=1, keepdims=True)
centered = M - row_means

print("shape после вычитания:", centered.shape)
print("сумма по строкам:", centered.sum(axis=1))
```
